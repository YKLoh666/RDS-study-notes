```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = [
    ("Coffee", 4),
    ("Tea", 2),
    ("Juice", 3),
    ("Coffee", 5),
    ("Soda", 3),
    ("Coffee", 5),
    ("Soda", 4),
    ("Juice", 2),
    ("Soda", 4),
    ("Coffee", 3),
]

df = pd.DataFrame(data, columns=["Favourite Drink", "Satisfaction Score"])
```

```python
plt.figure(figsize=(8, 5))

df_grouped = df.groupby("Favourite Drink")["Satisfaction Score"].mean().sort_values()
# horizontal bar chart
df_grouped.plot(kind="barh")
plt.title("Average Satisfaction Score by Favourite Drink")
plt.xlabel("Average Satisfaction Score")
plt.ylabel("")
plt.xlim(0, 5)
for index, value in df_grouped.items():
    y = float(df_grouped.index.get_loc(index))
    plt.text(value, y, f"{value:.2f}", va="center")
plt.show()
```
